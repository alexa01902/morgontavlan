```python
"""
FÖRELÄSNING 1 (forts.) - KLASSER (OOP)
Detta kodstycke visar: klasser, __init__, attribut, metoder, och hur man
skapar flera objekt (instanser) av samma klass
"""


# ============================================
# 1. EN ENKEL KLASS
# ============================================
# En klass är en "ritning" för att skapa objekt. Objektet kallas en "instans"
# __init__ är en specialmetod som körs automatiskt när ett nytt objekt skapas

class Student:
    def __init__(self, namn, alder):
        self.namn = namn      # attribut - information som hör till objektet
        self.alder = alder    # attribut

print("--- Enkel klass ---")
student1 = Student("Alexandra", 29)   # vi skapar en instans av klassen
print("Namn:", student1.namn)
print("Ålder:", student1.alder)
print()


# ============================================
# 2. FLERA INSTANSER AV SAMMA KLASS
# ============================================
# Varje instans har sina egna attribut, oberoende av andra instanser

print("--- Flera instanser ---")
student2 = Student("Erik", 34)
student3 = Student("Maria", 22)

for s in [student1, student2, student3]:
    print(f"{s.namn} är {s.alder} år gammal")
print()


# ============================================
# 3. METODER
# ============================================
# En metod är en funktion som "hör till" klassen och kan använda
# objektets egna attribut via "self"

class Student:
    def __init__(self, namn, alder):
        self.namn = namn
        self.alder = alder

    def presentera_sig(self):
        print(f"Hej, jag heter {self.namn} och är {self.alder} år.")

print("--- Metoder ---")
student1 = Student("Alexandra", 29)
student1.presentera_sig()
print()


# ============================================
# 4. METODER SOM ÄNDRAR ATTRIBUT
# ============================================
# Metoder kan också uppdatera objektets egna attribut

class Student:
    def __init__(self, namn, alder):
        self.namn = namn
        self.alder = alder
        self.godkanda_kurser = []   # tom lista från start

    def presentera_sig(self):
        print(f"Hej, jag heter {self.namn} och är {self.alder} år.")

    def lagg_till_kurs(self, kurs):
        self.godkanda_kurser.append(kurs)
        print(f"{self.namn} har nu klarat kursen: {kurs}")

print("--- Metoder som ändrar attribut ---")
student1 = Student("Alexandra", 29)
student1.lagg_till_kurs("Python för nybörjare")
student1.lagg_till_kurs("Python fortsättningskurs")
print("Godkända kurser:", student1.godkanda_kurser)
print()


# ============================================
# 5. KOMBINERAT EXEMPEL - ETT LITET KURSREGISTER
# ============================================
# Vi väver ihop klass, flera instanser, metoder och en loop

class Student:
    def __init__(self, namn, alder):
        self.namn = namn
        self.alder = alder
        self.godkanda_kurser = []

    def presentera_sig(self):
        print(f"Hej, jag heter {self.namn} och är {self.alder} år.")

    def lagg_till_kurs(self, kurs):
        self.godkanda_kurser.append(kurs)

    def antal_kurser(self):
        return len(self.godkanda_kurser)

print("--- Kombinerat exempel: kursregister ---")

# Skapar tre studenter (instanser)
alexandra = Student("Alexandra", 29)
erik = Student("Erik", 34)
maria = Student("Maria", 22)

# Lägger till kurser för var och en
alexandra.lagg_till_kurs("Python för nybörjare")
alexandra.lagg_till_kurs("Python fortsättningskurs")
erik.lagg_till_kurs("Python för nybörjare")

# Samlar alla studenter i en lista och loopar igenom dem
alla_studenter = [alexandra, erik, maria]

for student in alla_studenter:
    student.presentera_sig()
    print(f"Antal godkända kurser: {student.antal_kurser()}")
    print()
```

```python
import time

def tidtagning(base_func):
    def wrapper(n):
        start = time.time()
        resultat = base_func(n)
        slut = time.time()
        print(f"{base_func.__name__} tog {slut - start:.4f} sekunder")
        return resultat
    return wrapper

@tidtagning
def räkna_summa(n):
    total = 0
    for i in range(n):
        total += i
    return total

räkna_summa(10_000_000)
```

```python
import time

def tidtagning(base_func):
    def wrapper(*args, **kwargs):
        start = time.time()
        resultat = base_func(*args, **kwargs)
        slut = time.time()
        print(f"{base_func.__name__} tog {slut - start:.4f} sekunder")
        return resultat
    return wrapper

@tidtagning
def räkna_summa(n):
    total = 0
    for i in range(n):
        total += i
    return total

räkna_summa(10_000_000)
```

```python
# En vanlig funktion
def kvadrera(x):
    return x * x

kvadrera(5)   # 25

# Samma sak som en lambda
kvadrera = lambda x: x * x

kvadrera(5)   # 25
```

```python
# sorted() med key
namn = ["Anna", "Bo", "Kristina"]
sorted(namn, key=lambda x: len(x))
# ['Bo', 'Anna', 'Kristina']

# filter() – behåll bara det som stämmer
tal = [1, 2, 3, 4, 5, 6]
jämna = list(filter(lambda x: x % 2 == 0, tal))
# [2, 4, 6]

# map() – applicera på varje element
kvadrater = list(map(lambda x: x * x, tal))
# [1, 4, 9, 16, 25, 36]
```
