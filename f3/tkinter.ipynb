```python
class Student: # En klass för att skapa studentobjekt
    def __init__(self, namn, betyg): # Konstruktor som tar emot namn och betyg som argument
        self.namn = namn
        self.betyg = betyg

    def godkand(self): # Metod som returnerar True om betyget är 3 eller högre, annars False
        return self.betyg >= 3

elev1 = Student("Emma", 4) # Skapar ett studentobjekt med namn "Emma" och betyg 4
elev2 = Student("Lars", 2) # Skapar ett studentobjekt med namn "Lars" och betyg 2

print(elev1.godkand())  # anropar metoden godkand() för elev1 och skriver ut resultatet
print(elev2.godkand())  
```

```python
import tkinter as tk
from datetime import datetime

class MorgontavlanApp:
    def __init__(self, fonster):
        self.fonster = fonster
        self.fonster.title("Morgontavlan")
        self.fonster.configure(bg="#fcfdfd")

        self.klocka_etikett = tk.Label(self.fonster, text="", font=("Arial", 24),
                                        bg="#cde6ff", fg="black")
        self.klocka_etikett.pack()

        self.uppdatera_klocka()

    def uppdatera_klocka(self):
        nu = datetime.now() # Hämtar den aktuella tiden
        text = nu.strftime("%H:%M\n%A %d %B %Y") # Formaterar tiden till en sträng
        self.klocka_etikett.config(text=text)
        self.fonster.after(1000, self.uppdatera_klocka) # Uppdaterar klockan varje sekund

if __name__ == "__main__":
    fonster = tk.Tk()
    app = MorgontavlanApp(fonster)
    fonster.mainloop()
```

- lagg_till_uppgift: kollar att fältet inte är tomt (if uppgift != "") innan den läggs till, annars kan man klicka "Lägg till" på ett tomt fält och få tomma rader
- self.todo_falt.delete(0, tk.END) rensar textfältet efter att uppgiften lagts till, annars ligger den gamla texten kvar
- ta_bort_uppgift: curselection() returnerar en tuple (kan vara tom om inget är markerat), därför kollar vi if markerad: innan vi använder markerad[0]

```python
import tkinter as tk
from datetime import datetime

VECKODAGAR = {
    "Monday": "Måndag", "Tuesday": "Tisdag", "Wednesday": "Onsdag",
    "Thursday": "Torsdag", "Friday": "Fredag", "Saturday": "Lördag", "Sunday": "Söndag"
}

MANADER = {
    "January": "januari", "February": "februari", "March": "mars", "April": "april",
    "May": "maj", "June": "juni", "July": "juli", "August": "augusti",
    "September": "september", "October": "oktober", "November": "november", "December": "december"
}

# Färgpalett — mjuka, sammanhållna pastellfärger
BAKGRUND = "#f5f7fa"
FARG_KLOCKA = "#dceefb"
FARG_TODO = "#e3f5e5"
FARG_VADER = "#dceefb"
FARG_KLADTIPS = "#e6e3f7"
FARG_TRAFIK = "#faf1de"
FARG_SOMN = "#fbe4ee"
FARG_HUMOR = "#e3f5e5"
FARG_CITAT = "#f0f0f0"
ACCENT = "#4a90d9"


class MorgontavlanApp:
    def __init__(self, fonster):
        self.fonster = fonster
        self.fonster.title("Morgontavlan")
        self.fonster.configure(bg=BAKGRUND)

        # ------------ Klocka och datum ------------
        self.klock_ram = self.skapa_sektion(self.fonster, 0, 0, FARG_KLOCKA, "")

        self.tid_etikett = tk.Label(self.klock_ram, text="", font=("Arial", 36, "bold"), bg=FARG_KLOCKA)
        self.tid_etikett.grid(row=1, column=0, columnspan=2)

        self.datum_etikett = tk.Label(self.klock_ram, text="", font=("Arial", 14), bg=FARG_KLOCKA)
        self.datum_etikett.grid(row=2, column=0, columnspan=2)

        self.uppdatera_klocka()

        # ------------ Att-göra-lista ------------
        self.todo_ram = self.skapa_sektion(self.fonster, 1, 0, FARG_TODO, "Att-göra lista")

        self.todo_falt = tk.Entry(self.todo_ram, width=25, bg="white",
                                   relief="flat", highlightthickness=1,
                                   highlightbackground="#cccccc")
        self.todo_falt.grid(row=1, column=0, padx=(0, 8), ipady=4)

        self.lagg_till_knapp = tk.Button(self.todo_ram, text="Lägg till",
                                          command=self.lagg_till_uppgift,
                                          bg=ACCENT, fg="white", relief="flat", padx=10, pady=5)
        self.lagg_till_knapp.grid(row=1, column=1)

        self.todo_lista = tk.Listbox(self.todo_ram, width=40, bg="white",
                                      relief="flat", highlightthickness=1,
                                      highlightbackground="#cccccc")
        self.todo_lista.grid(row=2, column=0, columnspan=2, pady=10)

        self.ta_bort_knapp = tk.Button(self.todo_ram, text="Ta bort markerad",
                                        command=self.ta_bort_uppgift,
                                        bg="#e57373", fg="white", relief="flat", padx=10, pady=5)
        self.ta_bort_knapp.grid(row=3, column=0, columnspan=2)

        # ------------ Väder ------------
        self.vader_ram = self.skapa_sektion(self.fonster, 0, 1, FARG_VADER, "Väder")

        self.ort_etikett = tk.Label(self.vader_ram, text="Ort:", font=("Arial", 14), bg=FARG_VADER)
        self.ort_etikett.grid(row=1, column=0, sticky="w", pady=(0, 10))

        self.vader_falt = tk.Entry(self.vader_ram, bg="white", font=("Arial", 14))
        self.vader_falt.grid(row=1, column=1, pady=(0, 10))
        self.vader_falt.insert(0, "Skriv in ort")

        # ------------ Dagens klädtips ------------
        self.kladtips_ram = self.skapa_sektion(self.fonster, 0, 2, FARG_KLADTIPS, "Dagens klädtips")

        self.kladtips_etikett = tk.Label(self.kladtips_ram, text="Kyligt, ha många lager",
                                          font=("Arial", 14), bg=FARG_KLADTIPS)
        self.kladtips_etikett.grid(row=1, column=0, sticky="w", pady=(0, 10))

        # ------------ Kollektivtrafiken ------------
        self.trafik_ram = self.skapa_sektion(self.fonster, 1, 1, FARG_TRAFIK, "Kollektivtrafiken")

        # ------------ Sömn, humör och citat ------------
        self.bas_ram = tk.Frame(self.fonster, bg=BAKGRUND)
        self.bas_ram.grid(row=1, column=2, padx=20, pady=(10, 20))

        self.somn_ram = self.skapa_sektion(self.bas_ram, 0, 0, FARG_SOMN, "Sömn")
        self.somn_etikett = tk.Label(self.somn_ram, text="Hur många timmar sov du?",
                                      font=("Arial", 14), bg=FARG_SOMN)
        self.somn_etikett.grid(row=1, column=0, sticky="w", pady=(0, 10))

        self.humor_ram = self.skapa_sektion(self.bas_ram, 0, 1, FARG_HUMOR, "Morgonhumör")
        self.humor_etikett = tk.Label(self.humor_ram, text="Hur känner du dig idag?",
                                       font=("Arial", 14), bg=FARG_HUMOR)
        self.humor_etikett.grid(row=1, column=0, sticky="w", pady=(0, 10))

        self.citat_ram = self.skapa_sektion(self.bas_ram, 1, 0, FARG_CITAT, "Dagens citat")
        self.citat_ram.grid(columnspan=2)
        self.citat_etikett = tk.Label(self.citat_ram, text="Varje dag är en ny möjlighet.",
                                       font=("Arial", 14), bg=FARG_CITAT)
        self.citat_etikett.grid(row=1, column=0, sticky="w", pady=(0, 10))

    def skapa_sektion(self, förälder, rad, kolumn, bg, titel):
        """Skapar en ram med en fetstil-rubrik högst upp. Returnerar ramen
        så att fler widgets kan läggas till i den från rad 1 och nedåt."""
        ram = tk.Frame(förälder, bg=bg, padx=30, pady=20)
        ram.grid(row=rad, column=kolumn, padx=20, pady=(10, 20))

        if titel:
            rubrik = tk.Label(ram, text=titel, font=("Arial", 16, "bold"), bg=bg)
            rubrik.grid(row=0, column=0, columnspan=2, pady=(0, 10))

        return ram

    def uppdatera_klocka(self):
        nu = datetime.now()
        veckodag = VECKODAGAR[nu.strftime("%A")]
        manad = MANADER[nu.strftime("%B")]

        self.tid_etikett.config(text=nu.strftime("%H:%M:%S"))
        self.datum_etikett.config(text=f"{veckodag} {nu.day} {manad} {nu.year}")

        self.fonster.after(1000, self.uppdatera_klocka)

    def lagg_till_uppgift(self):
        uppgift = self.todo_falt.get()
        if uppgift != "":
            self.todo_lista.insert(tk.END, uppgift)
            self.todo_falt.delete(0, tk.END)

    def ta_bort_uppgift(self):
        markerad = self.todo_lista.curselection()
        if markerad:
            self.todo_lista.delete(markerad[0])


if __name__ == "__main__":
    fonster = tk.Tk()
    app = MorgontavlanApp(fonster)
    fonster.mainloop()
```
