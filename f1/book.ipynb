## Datatyper

Python har flera inbyggda datatyper. Här är de viktigaste:

```python
heltal = 25                # int (heltal)
decimaltal = 3.14          # float (decimaltal)
text = "Hej, Python!"      # str (sträng/text)
sanningsvarde = True       # bool (sant/falskt)
lista = [1, 2, 3, 4, 5]    # list (lista med flera värden)
```

Vi kan använda type() för att se vilken dartatypen variabel har

```python
print("--- Datatyper ---")
print(heltal, "är av typen", type(heltal))
print(decimaltal, "är av typen", type(decimaltal))
print(text, "är av typen", type(text))
print(sanningsvarde, "är av typen", type(sanningsvarde))
print(lista, "är av typen", type(lista))

 
```

## If / else - villkorssatser

Vi låter programmet fatta beslut baserat på ett värde

```python
alder = 20
 
print("--- If/else ---")
if alder < 18:
    print("Du är minderårig.")
elif alder < 65:
    print("Du är vuxen.")
else:
    print("Du är pensionär.")
```

## For-loop
En for-loop upprepar kod för varje element i något (t.ex. en lista)

```python
print("--- For-loop ---")
frukter = ["äpple", "banan", "apelsin"]
 
for frukt in frukter:
    print("Jag gillar", frukt)
 
# Vi kan också loopa igenom en siffersekvens med range()
print("\nRäknar från 1 till 5:")
for i in range(1, 6):
    print(i)
```

## While-loop
En while-loop upprepar kod så länge ett villkor är sant

```python
print("--- While-loop ---")
raknare = 0
 
while raknare < 3:
    print("Raknaren är just nu:", raknare)
    raknare += 1   # samma som raknare = raknare + 1
 
print("Loopen är klar, raknaren slutade på:", raknare)
```

## Importera moduler
Vi importerar 'random' för att kunna slumpa tal 

```python
import random
```

## Ett litet exempel

```python
print("--- Gissa talet ---")
 
hemligt_tal = random.randint(1, 10)  # slumpar ett tal mellan 1 och 10
gissningar = [4, 7, 2, 9, 5]          # en "spelares" gissningar (simulerat)
 
for gissning in gissningar:
    if gissning == hemligt_tal:
        print(f"Gissning {gissning}: Rätt! Du gissade {hemligt_tal}.")
        break
    elif gissning < hemligt_tal:
        print(f"Gissning {gissning}: För lågt.")
    else:
        print(f"Gissning {gissning}: För högt.")
```
