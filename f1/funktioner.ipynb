## Skapa en funktion

En funktion är ett återanvändbart kodblock. Vi definierar den med "def"
 och kallar på (anropar) den med dess namn + parenteser

```python
def hälsa():
    print("Hej och välkommen till kursen!")
 
print("--- Enkel funktion ---")
hälsa()      # här "anropar" vi funktionen
hälsa()      # vi kan anropa den flera gånger utan att skriva om koden
```

## Funktioner med paprametrar
Vi kan skicka in information till funktionen via parametrar.

```python
def hälsa_person(namn):
    print(f"Hej {namn}, kul att du är här!")
 
print("--- Funktion med parameter ---")
hälsa_person("Alexandra")
hälsa_person("Lars")
```

## Default-värden
Vi kan ge en parameter ett standardvärde ifall inget anges vid anropet.

```python
def hälsa_med_titel(namn, titel="student"):
    print(f"Hej {titel} {namn}!")
 
print("--- Default-värden ---")
hälsa_med_titel("Lindqvist")                  # använder default-titeln
hälsa_med_titel("Andersson", "professor")     # skriver över defaulten
```

## Returvärden
En funktion kan skicka tillbaka ett värde med "return".
Skillnaden mot print(): return ger oss ett värde vi kan spara och använda vidare.
 

```python
def addera(a, b):
    resultat = a + b
    return resultat
 
print("--- Returvärden ---")
summa = addera(5, 10)          # vi sparar det som returneras i en variabel
print("Summan är:", summa)
 
# Vi kan använda returvärdet direkt i annan kod, t.ex. i en till funktion
print("Dubbla summan är:", addera(summa, summa))
```

## Logik i funktioner
Funktioner kan innehålla vilken kod som helst, t.ex. if/else.

```python
def kontrollera_alder(alder):
    if alder < 18:
        return "minderårig"
    elif alder < 65:
        return "vuxen"
    else:
        return "pensionär"
 
print("--- Funktion med logik ---")
print("20 år gammal är:", kontrollera_alder(20))
print("70 år gammal är:", kontrollera_alder(70))
```

## Exempel - Temperaturomvandlare


```python
def celsius_till_fahrenheit(celsius):
    fahrenheit = celsius * 9 / 5 + 32
    return round(fahrenheit, 1)
 
print("--- Exempel: temperaturomvandlare ---")
temperaturer_celsius = [0, 20, 37, 100]
 
for temp in temperaturer_celsius:
    resultat = celsius_till_fahrenheit(temp)
    print(f"{temp}°C motsvarar {resultat}°F")
 
```

## Övning - Prisberäknare med mängdrabatt och medlemskap

Beskrivning an funktion:

Tre parametrar:
 - pris_per_styck  (obligatorisk)
 - antal           (obligatorisk)
 - medlem          (valfri, default = False)

Logiken:
  - Köper man 10 eller fler får man 10% mängdrabatt
  - Köper man 20 eller fler får man 20% mängdrabatt
  - Är man medlem får man dessutom ytterligare 5% rabatt på slutpriset

Retunera:
- Det beräknade slutpriset med två decimaler

```python
def berakna_totalpris(pris_per_styck, antal, medlem=False):
    totalpris = pris_per_styck * antal
 
    # Mängdrabatt baserat på antal
    if antal >= 20:
        totalpris = totalpris * 0.80   # 20% rabatt
        print(f"Mängdrabatt: 20% (köpte {antal} st)")
    elif antal >= 10:
        totalpris = totalpris * 0.90   # 10% rabatt
        print(f"Mängdrabatt: 10% (köpte {antal} st)")
    else:
        print(f"Ingen mängdrabatt (köpte {antal} st)")
 
    # Extra medlemsrabatt
    if medlem:
        totalpris = totalpris * 0.95   # ytterligare 5% rabatt
        print("Medlemsrabatt: 5%")
 
    return round(totalpris, 2)
```

```python
print("--- Scenario 1: Vanlig kund, litet köp ---")
pris = berakna_totalpris(pris_per_styck=50, antal=5)
print(f"Totalpris: {pris} kr\n")
 
print("--- Scenario 2: Vanlig kund, stort köp (mängdrabatt) ---")
pris = berakna_totalpris(pris_per_styck=50, antal=15)
print(f"Totalpris: {pris} kr\n")
 
print("--- Scenario 3: Medlem, stort köp (mängdrabatt + medlemsrabatt) ---")
pris = berakna_totalpris(pris_per_styck=50, antal=25, medlem=True)
print(f"Totalpris: {pris} kr\n")
 
print("--- Scenario 4: Medlem, litet köp (bara medlemsrabatt) ---")
pris = berakna_totalpris(50, 3, True)   # samma sak, utan namngivna argument
print(f"Totalpris: {pris} kr\n")
```

##  ANVÄND FUNKTIONEN PÅ EN LISTA MED BESTÄLLNINGAR

```python
print("--- Flera beställningar i en lista ---")
 
bestallningar = [
    {"pris": 120, "antal": 8, "medlem": False},
    {"pris": 120, "antal": 12, "medlem": False},
    {"pris": 120, "antal": 12, "medlem": True},
]
 
for i, order in enumerate(bestallningar, start=1):
    print(f"\nBeställning {i}:")
    totalpris = berakna_totalpris(order["pris"], order["antal"], order["medlem"])
    print(f"Totalpris: {totalpris} kr")
```
